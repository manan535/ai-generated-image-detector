## 1 · Configuration

All paths and model names live in `config.yaml` so the notebook
runs identically locally, on Colab, or on Kaggle — no hardcoded strings.

```python
import yaml, os, sys

# Allow imports from the project root
ROOT = os.path.abspath('..')
if ROOT not in sys.path:
    sys.path.insert(0, ROOT)

with open('../config.yaml') as f:
    config = yaml.safe_load(f)

print('=== config.yaml ===')
print(f"  data.cifake_path      : {config['data']['cifake_path']}")
print(f"  data.batch_size       : {config['data']['batch_size']}")
print(f"  data.num_workers      : {config['data']['num_workers']}")
print(f"  model.clip_model_name : {config['model']['clip_model_name']}")
print()
print('Dataset note: CIFAKE is used as the LOCAL smoke-test/demo dataset only.')
print('GenImage (~2.68 M images, 8 generators) is the primary training dataset')
print('and will be used during full training on cloud compute (Colab / Kaggle).')
```

## 2 · Dataset Loading

The loader is idempotent: it checks whether CIFAKE already exists on disk
before attempting a Kaggle download.  Running this cell twice is safe.

```python
import os
from torchvision.datasets import ImageFolder
from src.data.dataset import download_cifake

cifake_path = config['data']['cifake_path']
download_cifake(cifake_path)

# Load the raw splits to inspect them before any transforms
train_raw = ImageFolder(os.path.join(cifake_path, 'train'))
test_raw  = ImageFolder(os.path.join(cifake_path, 'test'))

print('=== Dataset loaded ===')
print(f'  Total images (train + test folders) : {len(train_raw) + len(test_raw):,}')
print()
# Read the class mapping from ImageFolder directly — never assume FAKE=0, REAL=1
print('  Class mapping (from ImageFolder):  class name → integer label')
for name, idx in train_raw.class_to_idx.items():
    print(f'    {idx}  →  {name}')
```

## 3 · Stratified Train / Val / Test Split (70 / 15 / 15)

The split is stratified on the real/fake label with a fixed seed (42)
so it is reproducible.  Sizes are computed from the actual data, not hardcoded.

```python
from torch.utils.data import ConcatDataset
from sklearn.model_selection import train_test_split
from collections import Counter

full_ds = ConcatDataset([train_raw, test_raw])
targets  = train_raw.targets + test_raw.targets
indices  = list(range(len(full_ds)))

train_idx, temp_idx, train_targets, temp_targets = train_test_split(
    indices, targets, test_size=0.30, stratify=targets, random_state=42
)
val_idx, test_idx, val_targets, test_targets = train_test_split(
    temp_idx, temp_targets, test_size=0.50, stratify=temp_targets, random_state=42
)

total = len(full_ds)
print('=== Dataset split ===')
print(f'  Total : {total:,} images')
print(f'  Train : {len(train_idx):,}  ({100*len(train_idx)/total:.1f}%)')
print(f'  Val   : {len(val_idx):,}   ({100*len(val_idx)/total:.1f}%)')
print(f'  Test  : {len(test_idx):,}   ({100*len(test_idx)/total:.1f}%)')
```

## 4 · Class Balance Check

IAD-B4 acceptance criteria: apply correction only if ratio is worse than 60/40.
CIFAKE is balanced by design; we verify this explicitly rather than assuming it.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# Read class names from ImageFolder to avoid hardcoding
idx_to_class = {v: k for k, v in train_raw.class_to_idx.items()}

splits = {
    'Train': Counter(train_targets),
    'Val':   Counter(val_targets),
    'Test':  Counter(test_targets),
}

print('=== Class balance per split ===')
for split_name, counter in splits.items():
    total_in_split = sum(counter.values())
    parts = [f"{idx_to_class[k]}={v} ({100*v/total_in_split:.1f}%)" for k, v in sorted(counter.items())]
    print(f'  {split_name:5s}: {", ".join(parts)}')

# Visual bar chart
labels   = list(idx_to_class.values())
x        = np.arange(len(splits))
width    = 0.35
fig, ax  = plt.subplots(figsize=(8, 4))

# Order bars by sorted label index
label_indices = sorted(idx_to_class.keys())
colors = ['#E57373', '#64B5F6']   # Red for FAKE, Blue for REAL

for i, label_idx in enumerate(label_indices):
    counts = [splits[s][label_idx] for s in splits]
    ax.bar(x + (i - 0.5) * width, counts, width,
           label=idx_to_class[label_idx], color=colors[i], alpha=0.85)

ax.set_xticks(x)
ax.set_xticklabels(list(splits.keys()), fontsize=12)
ax.set_ylabel('Number of images', fontsize=11)
ax.set_title('Class balance across splits (CIFAKE)', fontsize=13)
ax.legend(fontsize=11)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda val, _: f'{int(val):,}'))
plt.tight_layout()
plt.savefig('../outputs/class_balance.png', dpi=120)
plt.show()
print()
print('IAD-B4 decision: ratio is 50/50 — well within 60/40 threshold.')
print('Class-weighted loss and oversampling are NOT applied (intentional, not overlooked).')
```

## 5 · Preprocessing — Image → Model-Ready Tensor

Normalization stats are read from `CLIPImageProcessor` at runtime
(no hand-typed constants that could silently differ from ImageNet stats).

```python
import torch
from torchvision import transforms
from transformers import CLIPImageProcessor
from PIL import Image
import random, glob

processor = CLIPImageProcessor.from_pretrained(config['model']['clip_model_name'])
clip_mean  = processor.image_mean
clip_std   = processor.image_std

val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=clip_mean, std=clip_std)
])

# Pick one real image from CIFAKE
real_class = 'REAL'
sample_paths = glob.glob(f'{cifake_path}/train/{real_class}/*.jpg')
sample_path  = random.choice(sample_paths)
raw_img      = Image.open(sample_path).convert('RGB')
tensor       = val_test_transform(raw_img)

print('=== Preprocessing ===')
print(f'  Source file           : {os.path.basename(sample_path)}')
print(f'  Raw image size (PIL)  : {raw_img.size}  (W × H)')
print(f'  CLIP mean             : {clip_mean}')
print(f'  CLIP std              : {clip_std}')
print(f'  Tensor shape          : {tuple(tensor.shape)}  (C × H × W)')
print(f'  Tensor dtype          : {tensor.dtype}')
assert tensor.shape == (3, 224, 224), "Shape mismatch — preprocessing failed!"
print()
print('✓ Shape confirmed: 3 × 224 × 224')
```

## 6 · Augmentation Grid

These are **robustness** augmentations — they mimic real-world image degradations
(social-media JPEG re-compression, downscaling on upload, slight blur) that the
final model must handle.  They are applied to the **training split only**;
val and test remain on the clean resize+normalize pipeline.

| Augmentation | Range | p |
|---|---|---|
| JPEG recompression | quality 30 – 95 | 0.50 |
| Downscale → Resize | scale 0.5 – 0.9 | 0.50 |
| Gaussian Blur | kernel 3 – 7 | 0.30 |

```python
from IPython.display import Image as IPyImage, display

aug_grid_path = '../outputs/aug_grid.png'
assert os.path.exists(aug_grid_path), "aug_grid.png not found — run scripts/generate_aug_grid.py first"
display(IPyImage(filename=aug_grid_path, width=1100))
```

## 7 · DataLoader Validation

Pull one real batch from `train_loader` through the full augmented pipeline
and verify the shapes are exactly what the model will receive.

```python
# Use num_workers=0 for the notebook (avoids multiprocessing issues on Windows)
demo_config = dict(config)
demo_config['data'] = dict(config['data'])
demo_config['data']['num_workers'] = 0
demo_config['data']['batch_size']  = 8   # smaller batch so it's fast in the demo

from src.data.dataset import get_dataloaders
train_loader, val_loader, test_loader = get_dataloaders(demo_config)

images, labels = next(iter(train_loader))

print('=== DataLoader validation ===')
print(f'  Batch images shape : {tuple(images.shape)}   (batch × C × H × W)')
print(f'  Batch labels shape : {tuple(labels.shape)}')
print(f'  Labels in batch    : {labels.tolist()}')
print(f'  (0={idx_to_class[0]}, 1={idx_to_class[1]})')
print(f'  Images dtype       : {images.dtype}')
assert images.shape[1:] == (3, 224, 224), "Unexpected batch tensor shape!"
print()
print('✓ DataLoader batch shape confirmed: (8, 3, 224, 224)')
```

## 8 · CLIP ViT-B/16 — Feature Extraction Validation

We load the CLIP vision encoder and pass one preprocessed image through it
to confirm the 768-dimensional output we expect before fusion.

> **This is feature extraction validation only.**  
> No classifier head is attached. No training is done.

```python
from transformers import CLIPVisionModel

model_name   = config['model']['clip_model_name']
vision_model = CLIPVisionModel.from_pretrained(model_name)
vision_model.eval()

# Take one image from the batch
single_image = images[0].unsqueeze(0)   # shape: (1, 3, 224, 224)

with torch.no_grad():
    outputs       = vision_model(pixel_values=single_image)
    clip_features = outputs.pooler_output   # (1, 768)

print('=== CLIP ViT-B/16 feature extraction ===')
print(f'  Model             : {model_name}')
print(f'  Input shape       : {tuple(single_image.shape)}')
print(f'  Output shape      : {tuple(clip_features.shape)}')
print(f'  Feature dim       : {clip_features.shape[-1]}  (will be concatenated with 128-dim frequency branch → 896-dim fused vector)')
assert clip_features.shape == (1, 768), "Unexpected CLIP output shape!"
print()
print('✓ CLIP pooler_output confirmed: (1, 768)')
```
